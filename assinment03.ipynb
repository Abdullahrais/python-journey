Q3 Write a program in which use all the operators we can use in Python

Arithmetic Operators


```python
a=12
b=8
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponent:", a ** b)
```

comparison Operators

```python
print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

Assignment Operators

```python
c = 10
c += 5
print(c)

c -= 2
print(c)

c *= 2
print(c)

c /= 2
print(c)
```

 Logical Operators

```python
print(a > 5 and b < 10)
print(a > 20 or b < 10)
print(not(a == b))
```

Membership Operators

```python
numbers = [10, 20, 30]
print(20 in numbers)
print(40 not in numbers)
```

Identity Operators

```python
x = [1, 2]
y = x
z = [1, 2]

print(x is y)
print(x is not z)
```

Q4 Completes the following steps of small task: Mention Marks of English , Islamiat and Maths out of 100 in 3 different variables Mention Variable of Total Marks and assign 300 to it Calculate Percentage

```python
english = 85
islamiat = 90
maths = 95

total_marks = 300

obtained_marks = english + islamiat + maths

percentage = (obtained_marks / total_marks) * 100

print("English Marks:", english)
print("Islamiat Marks:", islamiat)
print("Maths Marks:", maths)
print("Obtained Marks:", obtained_marks)
print("Total Marks:", total_marks)
print("Percentage:", percentage, "%")
```
