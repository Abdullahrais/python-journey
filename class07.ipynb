```python
def calc_percentage_no_parameter():
    bio = int(input("enter obtained marks :"))
    chem = int(input("enter obtained marks :"))
    phy = int(input("enter obtained marks :"))
    math = int(input("enter obtained marks :"))
    total = 400
    obtained = bio + chem + phy + math
    percentage = (obtained / total) * 100
    print(f"Percentage is {percentage}%")
```

```python
calc_percentage_no_parameter()
```

```python
#Parameter wo variable hota hai jo function banate waqt brackets () ke andar likha jata hai.    
def calculating_percentage(bio, chem, engl, phy):
    total_marks = 400
    obtained_marks = bio + chem + engl + phy
    percentage =  (obtained_marks * 100)/total_marks
    print(f"PERCENTAGE IS {percentage}")
```

```python
# The number in the brackets are arguments
calculating_percentage(23, 45, 24, 65)
```

```python
bio = int(input("enter obtained marks :"))
chem = int(input("enter obtained marks :"))
phy = int(input("enter obtained marks :"))
math = int(input("enter obtained marks :"))
calculating_percentage(bio, chem, phy, math)
```
